## __Fundamental Go__

Go memiliki alur bekerja seperti berikut ini:

```text

Source Code
     ↓
Go Compiler
     ↓
Type Checking
     ↓
Machine Code / Binary
     ↓
OS menjalankan program

```

Go adalah compiled, __statically typed language__.

contoh:


`````go
package main

import "fmt"

func main() {
    name := "Ben"
    fmt.Println(name)
}
`````

jadi ketika ```name := "Ben"``` compiler sudah tau bahwa 

- variable nya bernama ````name```
- value = ```"ben"```
- type nya adalah __string__


Golang __bukan bahasa pemograman yang bersifat dynamic__. ketika suatu varible yang dibentuk sudah di deklarasikan type-nya atas suatu value, maka type-nya seterunya akan bersifat sama, tidak bisa berubah-berubah

__contoh:__
```go
// ketika age = int
age := 19 
// maka golang akan selalu mengingat variable age itu akan menyimpan nilai int / angka

// ketika age dicoba isi string
age := "nineteen" 
// maka akan terjadi error compiler
```

> __Ingat__ <br/>
meskipun kita bisa menulis type data di go secara tidak eksplisit dengan menggunakan ```:=```, namun pada hakikatnya __GO bukanlah tipe bahasa yang dynamic typing__ seperti javascript atau python, <br/> sehingga apapun yang sudah di deklarasikan diawal, tidak bisa di ubah lagi. Penggunaan ```:=``` hanyalah sebagai __type inference dan hal itu tidak sama dengan dynamic typing__.

## __Variable, Type, dan Zero Value__

ada beberapa cara untuk membuat variable dalam bahasa golang 

#### __cara 1__
menggunakan zero value, dimana hanya menginisiasi variable saja tapi value nya belum, sehingga GO tidak membiarkannya memiliki garbage value, sehingga pada akhirnya nilainya otomatis ```0```

__contoh__ :
```go
// hanya inisiasi variable saja
var age int
// maka nilainya = 0
```


#### __cara 2__
tidak membiarkan variablenya kosong, melainkan memberi value langsung ke variabel yang di assign

__contoh__ :
```go
// hanya inisiasi variable saja
var age int = 2007
// maka nilainya = 0
```


#### __cara 3__
menggunakan compiler inference dan ini biasanya sering di pakai untuk local variabel dengan ciri khasnya ```:=```

__contoh__ :
```go
// hanya inisiasi variable saja
 age := 2007
// maka nilainya = 0
```


### __Zero Value__

contoh-contoh zero value dari masing-masing jenis type yang ada 

| Type      | Zero Value |
| --------- | ---------- |
| `int`     | `0`        |
| `float64` | `0`        |
| `string`  | `""`       |
| `bool`    | `false`    |
| pointer   | `nil`      |
| slice     | `nil`      |
| map       | `nil`      |
| channel   | `nil`      |
| function  | `nil`      |
| interface | `nil`      |


__contohnya__ : 

`````go
package main

import "fmt"

var age int
var name string
var active bool

fmt.Println(age)
fmt.Println(name)
fmt.Println(active)
`````

#### __Kenapa backend engineer perlu memahami zero value?__

Misalnya:

```go
type User struct {
    Name     string
    Age      int
    IsActive bool
}
```

lalu:

```go
var user User
```

Maka sebenarnya yang ada dipakai si ```var user``` adalah type yang di deklarasikan di ```type User struct``` yang isinya ada:

```go
User{
    Name:     "",
    Age:      0,
    IsActive: false,
}
```

*asumsikan sebagai templatenya

Tidak ada undefined. Ini salah satu karakteristik Go yang membuat state lebih predictable.

#### __ingat !__

Ringkasan paling gampang:

* `nil` = variabelnya ada, tapi tempat penyimpanannya belum dibuat.
* `slice` yang masih `nil` **boleh di-append**, karena `append()` bisa otomatis membuat tempat penyimpanan.
* `map` yang masih `nil` **tidak boleh langsung diisi**, karena map-nya belum dibuat.
* Jadi map harus dibuat dulu pakai `make()`.

Contoh:

```go
var numbers []int
numbers = append(numbers, 10) // boleh
```

Sedangkan:

```go
var users map[string]string
users["1"] = "Ben" // panic / error
```

Harus:

```go
users := make(map[string]string)
users["1"] = "Ben" // boleh, karna dia dibuat `make` dlu
```

Cara ingatnya:

```text
Slice → append bisa bikin tempat sendiri
Map   → harus make tempat dulu
```

Kalau interview, cukup ingat:

> `nil slice` bisa di-append, tapi `nil map` harus diinisialisasi dengan `make()` sebelum ditulis.


## __Control Flow__

dalam bahasa Golang, ada beberapa mekanisme yang digunakan untuk control flow seperti:


### __1. IF__
```go
age := 19

if age >= 18 {
    fmt.Println("adult")
}
```

Tidak perlu:

```go
if (age >= 18)
```

sehingga Go tidak membutuhkan parentheses


### __2. If + initializer__

selain hanya sebatas If, GO juga punya flow control yang mengkombinasikan if dengan juga suatu __kondisi inisiasi__ seperti:

```go
if err := doSomething(); err != nil {
    return err
}
```

Ini penting sekali untuk backend

Contoh:

```go
func saveUser() error {
    return nil
}

func main() {
    if err := saveUser(); err != nil {
        fmt.Println(err)
        return
    }

    fmt.Println("success")
}
```

Variable: ```err``` hanya hidup dalam scope if. Ini disebut: __lexical scope__


### __3. FOR__

Go secara fundamental hanya punya: ```for```

Tidak ada keyword: ```while```

Contoh:

```go
for i := 0; i < 5; i++ {
    fmt.Println(i)
}
```

hasil:

```text
0
1
2
3
4
```
sementara itu, untuk while juga bisa diakalin dengan ```while-style```, contohnya: 

```go
// While-style
i := 0

for i < 5 {
    fmt.Println(i)
    i++
}
```

## __Function__

dalam implementasi function dalam bahsa go, terdapat beberapa cara penulisan yang kerap sering kali dijumpain, seperti: 

### __Function sederhana__:

```go
func add(a int, b int) int {
    return a + b
}
```

dari function diatas bisa dipersingkat menjadi :

```go
func add(a, b int) int {
    return a + b
}
```

Mekanisme function dalam go sama seperti fungsi pada umumnya dimana 

```Input``` > ```function``` > ```Output```

Dalam backend nanti bentuknya berkembang menjadi:

```HTTP Request``` > ```Handler``` > ```Service Function``` > ```Repository Function``` > ```Database```

Pada dasarnya tetap:

```Input``` > ```function``` > ```Output```

selain dari penulisan function pada umumnya, function di GO juga dapat ditulis sebagai:

### __Multiple Return Values__

Go punya fitur penting:

```go
func divide(a, b int) (int, error) {
    if b == 0 {
        return 0, fmt.Errorf("cannot divide by zero")
    }

    return a / b, nil
}
```

Pemakaiannya bisa seperti :

```go
result, err := divide(10, 2)

if err != nil {
    fmt.Println(err)
    return
}

fmt.Println(result)
```

Perhatikan pola:
```go
value, err := function()
```

Nanti dalam backend hampir setiap hari kamu akan melihat:

```go
user, err := repository.FindUser(...)
```

atau:

```go
rows, err := db.Query(...)
```
atau:

```go
data, err := json.Marshal(...)
```

Jadi __error handling Go sebenarnya dibangun sangat erat dengan function return values__

Array Go:
```go

```

```text
index
```

hasil: ```30```

adalah dua type berbeda.


Jadi: ```var a [3]int``` dan  ```var b [4]int```



a dan b bukan tipe yang sama. Kenapa? Karena panjang array merupakan bagian dari type.


```text

 ↑
length termasuk type
```
