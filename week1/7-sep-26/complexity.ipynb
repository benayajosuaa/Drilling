# __Complexity__

September 7, 2026 by __Benaya Joshua__ 

Berikut versi ringkasnya:

| Complexity   | Gambaran Sederhana                              | Contoh di Go                | Cara Ingat      |
| ------------ | ----------------------------------------------- | --------------------------- | --------------- |
| **O(1)**     | Kerja tetap, walau data makin banyak            | `numbers[2]`                | Ambil langsung  |
| **O(log n)** | Data dibagi 2 terus                             | Binary Search               | Potong setengah |
| **O(n)**     | Cek semua data satu per satu                    | `for _, x := range numbers` | 1 loop          |
| **O(n²)**    | Setiap data dibandingkan dengan semua data lain | loop di dalam loop          | 2 nested loop   |
| **O(n³)**    | Tiga loop bersarang                             | 3 nested loop               | 3 nested loop   |

### Contoh cepat

| Kode                     | Complexity   | Kenapa                |
| ------------------------ | ------------ | --------------------- |
| `numbers[0]`             | **O(1)**     | Langsung ambil 1 data |
| `for i := 0; i < n; i++` | **O(n)**     | Loop sebanyak `n`     |
| 2 loop terpisah          | **O(n)**     | `n + n = 2n → O(n)`   |
| loop di dalam loop       | **O(n²)**    | `n × n`               |
| `n = n / 2` dalam loop   | **O(log n)** | Data terus dibagi 2   |

Cara paling gampang menentukan complexity:

| Yang Dilihat                       | Kemungkinan Complexity |
| ---------------------------------- | ---------------------- |
| Tidak ada loop, langsung akses     | **O(1)**               |
| Ada 1 loop berdasarkan jumlah data | **O(n)**               |
| Ada loop di dalam loop             | **O(n²)**              |
| Ada 2 loop tetapi terpisah         | **O(n)**               |
| Data terus dibagi 2                | **O(log n)**           |

Intinya:

> **Complexity = melihat bagaimana jumlah pekerjaan program bertambah ketika jumlah data `n` semakin besar.**

Contoh kalau `n = 1.000`:

| Complexity | Kira-kira pekerjaan |
| ---------- | ------------------: |
| O(1)       |                   1 |
| O(log n)   |                 ~10 |
| O(n)       |               1.000 |
| O(n²)      |           1.000.000 |

Untuk sekarang, Ben, cukup kuatkan dulu **O(1), O(n), O(n²), dan O(log n)**. Empat ini paling penting buat drilling interview awal.
