### __Latihan 2b — Expense Classifier__

__Buat function__ :

func classifyExpense(amount int) string

<br/>

__Aturan__ :

|            Amount | Category |
| ----------------: | -------- |
|            `<= 0` | Invalid  |
|      `1 - 100000` | Small    |
| `100001 - 500000` | Medium   |
|        `> 500000` | Large    |

<br/>

__Contoh__ : ```classifyExpense(50000)```

<br/>

__Output__ : Small

<br/>

__Test Case Wajib__
```text
-10000   → Invalid
0        → Invalid
50000    → Small
100000   → Small
100001   → Medium
300000   → Medium
500000   → Medium
500001   → Large
1000000  → Large
```

````go
package main 

import "fmt"


func classifyExpense(amount int) string {
	if amount <= 0 {
		return "invalid"
	} else if amount <= 100000 {
		return "small"
	} else if amount <= 500000 {
		return "medium"
	} else {
		return "large"
	}
}


func main(){
	fmt.Println(classifyExpense(-10000))
	fmt.Println(classifyExpense(1000000))
	fmt.Println(classifyExpense(500001))
	fmt.Println(classifyExpense(500000))
}

main()
````

## __Function__

### __Latihan 3a — Function Sederhana__

__Buat__ :

```func add(a int, b int) int```

<br/>

__Contoh__ :
```go
result := add(10, 5)

fmt.Println(result)
```

<br/>

__Output__ : ```15```

Kemudian buat: ```subtract()```, ```multiply()```

````go
package main

import "fmt"


func add(a int, b int) int {
	return a + b
}

func subtract(a int, b int) int {
	return a - b
}

func multiply(a int, b int) int {
	return a * b
}

func divide(a, b int) float64 {
	if b == 0 {
		return 0
	}
	return float64(a) / float64(b)
}

func main() {
	x := 10
	y := 4

	fmt.Println("Hasil Add:", add(x, y))
	fmt.Println("Hasil Subtract:", subtract(x, y)) 
	fmt.Println("Hasil Multiply:", multiply(x, y)) 
	fmt.Println("Hasil Divide:", divide(x, y))     
}

main()
````

### __Latihan 4B — Total Array__

__Dari array yang sama__:

```go
scores := [5]int{
	80,
	70,
	90,
	60,
	100,
}
```

<br/>

__Hitung__ Total: 400


<br/>

__Hint__:

```go
total := 0

for ... {
	total += ...
}
```

````go
package main 

import "fmt"

func main(){

	scores := [5]int{80, 70, 90, 60, 100}

	total := 0

	for i := 0; i < len(scores); i++ {
		total += scores[i]
	}

	fmt.Println("Total skor:", total)
}


main()
````

### __Soal 4c — Analisis Nilai Mahasiswa__

__Gunakan__:
```text
scores := [8]int{
	80,
	40,
	90,
	55,
	100,
	30,
	75,
	60,
}
```

<br/>

__Cari__: Total, Average, Highest, Lowest, Passed, Failed

<br/>

__Aturan__:
* Nilai >= 60 → Passed
* Nilai < 60  → Failed

<br/>

__Expected result__:
```text
Highest : 100
Lowest  : 30
Passed  : 5
Failed  : 3
```

<br/>

Untuk Total dan Average, hitung sendiri. Batasan Jangan menggunakan library untuk: max, min, sum. Logic harus dibuat sendiri.

````go
package main 

import "fmt"

// scores := [8]int{80, 40, 90, 55, 100, 30, 75, 60,}

func pass(number [8]int) int{
	passed := 0
	for i := 0; i < len(number); i++ {
		if number[i] >= 60 {
			passed = passed + 1
		}
	}
	return passed
}

func failed(number [8]int) int{
	failed := 0
	for i := 0; i < len(number); i++ {
		if number[i] < 60 {
			failed = failed + 1
		}
	}
	return failed
}


func sum(number [8]int) int {
	total := 0
	for i := 0; i < len(number); i++ {
		total += number[i]
	}
	return total
}

func average(number [8]int) int {
	total := 0
	for i := 0; i < len(number); i++ {
		total += number[i]
	}
	return total / len(number)
}

func min(number [8]int) int {
	minVal := number[0]
	for i := 0; i< len(number); i++ {
		if number[i] < minVal {
			minVal = number[i]
		}
	}
	return minVal
}

func max(number [8]int) int {
	maxVal := number[0]
	for i := 0; i < len(number); i++ {
		if number[i] > maxVal {
			maxVal = number[i]
		}
	}
	return maxVal
}


func main() {
	scores := [8]int{80, 40, 90, 55, 100, 30, 75, 60,}
	


	hasilSum := sum(scores)
	hasilMin := min(scores)
	hasilMax := max(scores)
	hasilAverage := average(scores)

	passed := pass(scores)
	failed := failed(scores)

	fmt.Println("Total:", hasilSum) 
	fmt.Println("Min:", hasilMin) 
	fmt.Println("Max:", hasilMax) 
	fmt.Println("Average:", hasilAverage) 
	
	fmt.Println("Passed:", passed) 
	fmt.Println("Failed:", failed) 
}

main()
````
